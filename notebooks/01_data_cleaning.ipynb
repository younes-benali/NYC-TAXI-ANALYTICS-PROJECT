```python
import pandas as pd
import numpy as np
from pathlib import Path
from datetime import datetime
```

```python
# Paths
RAW_DIR = Path("../data/raw")
PROCESSED_DIR = Path("../data/processed")
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

print("Setup complete")
print(f"Raw data directory: {RAW_DIR}")
print(f"Processed directory: {PROCESSED_DIR}")
```

## Cleaning Function

```python
import sys
from pathlib import Path
sys.path.insert(0, str(Path.cwd().parent))   
from src.clean import clean_dataframe
```

### Data is being divided into four batches due to its excessive size (48 million rows).

### Cleaning Batch 1 (Jan–Mar)

```python
print("Cleaning BATCH 1: January, February, March 2025")
print("="*50)

# Loading months 01, 02, 03

months = ["01", "02", "03"]
df_list = []
for month in months:
    file = RAW_DIR / f"yellow_tripdata_2025-{month}.parquet"
    if file.exists():
        df = pd.read_parquet(file)
        df_list.append(df)
        print(f"Loaded {file.name}: {len(df):,} rows")
    else:
        print(f" Missing {file.name}")

if df_list:
    df_batch = pd.concat(df_list, ignore_index=True)
    print(f"\nTotal raw rows in batch: {len(df_batch):,}")
    
    # Cleaning
    df_clean = clean_dataframe(df_batch)
    print(f"Cleaned rows: {len(df_clean):,} ({len(df_clean)/len(df_batch)*100:.1f}% retained)")
    
    # Saving cleaned data
    output_file = PROCESSED_DIR / "cleaned_batch1_jan_mar.parquet"
    df_clean.to_parquet(output_file, index=False)
    print(f" Saved to {output_file}")
    
    # Free memory
    del df_batch, df_clean
else:
    print(" No data for batch 1")
```
